```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
import scipy
from scipy import stats

plt.rcParams["figure.figsize"]=(15,10) #set figure sizes
plt.rcParams["figure.dpi"]=300 #set dpi of figures (for clarity)
np.random.seed(42) #set random seed (so that it looks the same every time)
```

```python
#loc and scale values can be changed here
loc=3.14
scale=0.02
data=stats.norm.rvs(loc=loc,scale=scale,size=100000)
```

```python
#first plot (linear-linear histogram of the distribution with lines)
fig,ax=plt.subplots(1,1)
#blue histogram
ax.hist(data,50,density=True,color="b",alpha=0.7,histtype="stepfilled")
#set sizes of the numbers on the axes
plt.tick_params(labelsize=24)
#set limits on the x-axis
plt.xlim([loc-5*scale,loc+5*scale])
#create arrays with linear spaces from -5 sigmas to +5 sigmas
x=np.linspace(loc-5*scale,loc+5*scale,1000)
#red lines
ax.plot(x,stats.norm.pdf(x,loc=loc,scale=scale),linewidth=8,alpha=0.7,c="r")
#show plot
plt.show()
```

```python
#second plot (semi-log histogram of the distribution)
fig,ax=plt.subplots(1,1)
ax.hist(data,50,density=True,color="b",alpha=0.7,histtype="stepfilled")
plt.yscale("log")
plt.tick_params(labelsize=24)
plt.xlim([loc-5*scale,loc+5*scale])
plt.show()
```

```python
#third plot (semi-log histogram of the distribution with lines)
fig,ax=plt.subplots(1,1)
ax.hist(data,50,density=True,color="b",alpha=0.7,histtype="stepfilled")
plt.yscale("log")
plt.tick_params(labelsize=24)
plt.xlim([loc-5*scale,loc+5*scale])
ax.plot(x,stats.norm.pdf(x,loc=loc,scale=scale),linewidth=8,alpha=0.7,c="r")
plt.show()
```
